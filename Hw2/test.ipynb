```python
import nbformat

def truly_fix_widgets(filename):
    # Load the notebook
    nb = nbformat.read(filename, as_version=nbformat.NO_CONVERT)

    # Check if widgets metadata exists
    if 'widgets' in nb.metadata:
        widget_meta = nb.metadata['widgets']
        app_widget = widget_meta.get('application/vnd.jupyter.widget-state+json', {})
        
        # If "state" is missing but widget IDs exist
        if 'state' not in app_widget:
            new_state = {}
            # Move everything inside "state"
            for k, v in list(app_widget.items()):
                if k != 'state':
                    new_state[k] = v
                    del app_widget[k]
            app_widget['state'] = new_state

    # Save to a new clean file
    fixed_filename = filename.replace('.ipynb', '_fixed.ipynb')
    nbformat.write(nb, fixed_filename)
    print(f"✅ Fixed notebook saved to {fixed_filename}")

# Example usage
truly_fix_widgets('HW2_4_Sentence_Contrastive_Learning.ipynb')
```
